```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.preprocessing import StandardScaler,LabelEncoder
from sklearn.pipeline import Pipeline

from sklearn.svm import SVC
from sklearn.metrics import classification_report, confusion_matrix

import joblib
import matplotlib.pyplot as plt
```

```python
np.random.seed(42)
```

```python
n=1200
```

```python
legit_amount = np.random.normal(loc = 1200, scale = 600, size = n).clip(50,10000)
```

```python
len(legit_amount)
```

```python
legit_hours = np.random.randint(7,22, n)
```

```python
legit_hours
```

```python
legit_merchant_risk = np.random.normal(loc = 25,scale = 10, size = n).clip(1,70)
```

```python
legit_merchant_risk
```

```python
legit_user_avg = np.random.normal(loc = 3000, scale = 1500, size = n).clip(100,15000)
```

```python
fraud_amount = np.random.normal(loc = 4000, scale = 2000, size = n).clip(100,20000)
fraud_hours = np.random.choice(
    [0,1,2,3,4,5,6,23], size = n, p = [0.15,0.15,0.15,0.15,0.1,0.1,0.1,0.1]
)
fraud_merchant_risk = np.random.normal(loc = 70, scale = 15, size = n).clip(30,100)
fraud_user_avg = np.random.normal(loc = 800, scale = 400, size = n).clip(50,5000)
```

```python
x = np.vstack([
    np.column_stack([legit_amount, legit_hours, legit_merchant_risk, legit_user_avg]),
    np.column_stack([fraud_amount, fraud_hours, fraud_merchant_risk, fraud_user_avg])
])
```

```python
x
```

```python
y_labels= np.array(["legit"]*n+["Fraud"]*n)
```

```python
y_labels
```

```python
df = pd.DataFrame(x,columns=["amount_inr","txn_hour","merchant_risk_score","user_avg_txn"])
```

```python
df
```

```python
df["label"] = y_labels
```

```python
y_labels
```

```python
df
```

```python
x = df.drop("label", axis=1)
y = df["label"]
```

```python
x_train, x_test, y_train, y_test = train_test_split(
    x, y, test_size=0.2, random_state=42)
```

```python
base_model = Pipeline(steps=[
    ('svc', SVC(kernel='rbf',probability=True))
])
```

```python
base_model
```

```python
base_model.fit(x_train, y_train)
```

```python
base_model.predict(x_test)
```

```python
accuracy = base_model.score(x_test, y_test)
print(f"Model accuracy: {accuracy*100:.2f}%")
```

```python
grid = GridSearchCV(
    estimator=base_model,
    param_grid={
        'svc__C': [0.1, 1, 10, 100],
        'svc__gamma': [1, 0.1, 0.01, 0.001],
        'svc__kernel': ['rbf','linear'],
        'svc__degree': [2,3,4]
    },
    cv = 5,
    scoring='accuracy',
    verbose=2,
    n_jobs=-1
)
```

```python
grid.fit(x_train, y_train)
```

```python
best_model = grid.best_estimator_
```

```python
grid.best_params_
```

```python
grid.best_score_
```

```python
accuracy_score = best_model.score(x_test, y_test)
print(f"Optimized Model accuracy: {accuracy_score*100:.2f}%")
```

```python
joblib.dump(best_model, "svc_fraud_model.pkl")
```
