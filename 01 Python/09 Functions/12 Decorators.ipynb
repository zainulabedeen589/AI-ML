```python
def my_deco(func):
    def wrapper():
        print("print this before the function")
        func()
        print("do this after the fcuntion")
    return wrapper
```

```python
@my_deco
def greet():
    print("welcome to euron")
```

```python
greet()
```

```python
@my_deco
def welcome():
    print("welcome to euron")
```

```python
welcome()
```

```python
import time
def print_list(l):
    l1 = []
    start_time = time.time()
    for i in l :
        if i  % 2 == 0 :
            l1.append(i)
        time.sleep(1)
    end_time = time.time()
    total_time_taken = end_time - start_time
    return l1,total_time_taken
```

```python
print_list([4,5,6,7,8,9,6,67,7,8,89,6,6,7,8,9,34,4,5,6,76,78,8,98])
```

```python
import time 
def time_cal_deco(func):
    def calculator(*args,**kwargs):
        start = time.time()
        print(start)
        OUTPUT = func(*args,**kwargs)
        end = time.time()
        print(end) 
        total_time = end-start
        return OUTPUT , total_time
    return calculator
               
```

```python
@time_cal_deco
@my_deco
def print_list(l):
    l1 = []
    for i in l :
        if i  % 2 == 0 :
            l1.append(i)
    return l1
```
