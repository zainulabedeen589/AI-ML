```python
def add(a,b):
    return a+b
```

```python
add(4,5)
```

```python
c = lambda a,b : a+b
```

```python
c(4,5)
```

```python
square = lambda a : a**2
```

```python
square(7)
```

```python
max = lambda a,b : a if a>b else b
```

```python
max(5,7)
```

```python
even_odd = lambda number : "even" if number%2 == 0 else "odd"
```

```python
even_odd(45)
```

```python
even_odd(24)
```
