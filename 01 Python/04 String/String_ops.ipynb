```python
from string import Template
```

```python
Temp = Template("my name is $name , my age is $age , i am working with $company")
```

```python
temp = Temp.substitute(name=name, age=age, company=company)
```

```python
temp
```

```python
euron_email_temp = Template(
    """
    From:$sender
    To:$receiver
    Subject:$subject
    $body
    """
)
```

```python
result = euron_email_temp.substitute(
    sender=input("enter your email"),
    receiver="sskumar@gmail.com",
    subject="Welcome to Euron",
    body="Welcome to Euron, we are happy to have you in our family",
)
```
