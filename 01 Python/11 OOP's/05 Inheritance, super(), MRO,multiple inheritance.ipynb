```python
class course:    
    def __init__(self,title):
        self.title= title
        print(f"{self.title}")
```

```python
class euroncourse(course):
    def __init__(self,title, mentor):
        super().__init__(title=title)
        self.mentor = mentor
        print(f"{self.mentor},{self.title}")
```

```python
eu= euroncourse("python","sudh")
```
