```python
class cousre :
    def __init__(self,topics):
        self.topics = topics
        
    def __len__(self):
        return len(self.topics)
```

```python
c = cousre("sudh")
```

```python
len(c)
```

```python
c1 = cousre([2,3,4,5,6])
```

```python
len(c1)
```

```python
c1 = cousre((2,3,4,5,6))
```

```python
len(c1)
```
