```python
df['averagescore'] = (df['math score'] + df['reading score'] + df['writing score']) / 3
```

```python
df
```

```python
def math_performance(score):
    if score['math score'] >= 60:
        return 'good'
    elif score['math score'] >= 40:
        return 'average'
    else:
        return 'poor'

df['math_perfrmance'] = df.apply(math_performance, axis=1)
```

```python
df
```

```python
def newCol(score):
   return  score['race/ethnicity'].replace('group ','') 
df['newCol'] = df.apply(lambda score: newCol(score), axis=1)  
df.head()
```

```python
df['race_short'] = df['race/ethnicity'].str.replace('group ', '', case=False)
```

```python
df.head()
```

```python
def getGroups(row):
   l =  row['race/ethnicity']
   return l[:1]
df['group'] = df.apply(getGroups, axis=1)
df.head()
```

```python
def Group(group):
    if group['race/ethnicity'] == 'group A':
        return 'A'
    elif group['race/ethnicity'] == 'group B':
        return 'B'
    else:
        return 'C'
    
df['group'] = df.apply(Group, axis=1)
```

```python
df
```

```python
def getGroups(row):
   l =  row['race/ethnicity'].split(' ')
   return l[1]
df['group'] = df.apply(getGroups, axis=1)
df.head(10)
```

```python
df['gender_num'] = df['gender'].map({'female': 1,'male': 0})
```

```python
df.head()
```

```python
df['racecat'] = df['race/ethnicity'].map({'group A': "A", 'group B': "B", 'group C': "C", 'group D': "D", 'group E': "E"})
```
