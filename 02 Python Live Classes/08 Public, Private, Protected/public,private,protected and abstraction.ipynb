```python
from abc import ABC, abstractmethod
```

```python
class payment(ABC):
    @abstractmethod
    def process_payment(self,amount):
        pass
    
    @abstractmethod
    def payment_details(self):
        pass
    
    @abstractmethod
    def common_method(self):
        pass
    
    @abstractmethod
    def check_balance(self):
        pass
```

```python
a = payment()

a.process_payment(100)
```

```python
class paypal(payment):
    def process_payment(self,amount):
        print(f"Processing PayPal payment of {amount}")
        
    def payment_details(self):
        print("PayPal payment details")
        
    def common_method(self):
        print("Common method implementation in PayPal")
```

```python
p = paypal()
p.process_payment(100)
```

```python
p.payment_details()
```

```python
def my_decorator(func):
    def wrapper():
        print("its a part of euron")
        func()
        print("this is the end")
    return wrapper
```

```python
import time
@timer
@my_decorator
def timercalfunction():
    for i in range(3):
        print(f"Iteration {i+1} started")
    time.sleep(2)
    print("function executed")
```

```python
timercalfunction()
```

```python
def timer(func):
    def wrapper():
        start_time = time.time()
        func()
        end_time = time.time()
        print(f"Function executed in {end_time - start_time} seconds")
    return wrapper
```
