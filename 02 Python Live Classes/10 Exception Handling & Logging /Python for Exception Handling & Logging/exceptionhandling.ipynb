```python
try:
    print("this is my outer try block ")
    try:
        print("this is my inner try block")
        x = 1 / 0
    except ValueError as e:
        print("inner except block:", e)
    finally:
        print("inner finally block executed")
except Exception as e:
    print("outer except block:", e)
        
```

```python
try :
    print("welcome to calculator from euron")
    try :
        a = int(input("enter first number : "))
        b = int(input("enter second number : "))
        try :
            op = input("enter operator (+ , - , * , / ) : ")
            if op == "+" :
                print("addition is " , a + b)
            elif op == "-" :
                print("subtraction is " , a - b)
            elif op == "*" :
                print("multiplication is " , a * b)
            elif op == "/" :
                print("division is " , a / b)
            else :
                print("invalid operator")
        except Exception as e :
            print("inner most except block :" , e)
    except Exception as e :
        print("inner except block :" , e)
except Exception as e :
    print("outer except block :" , e)
finally:
    print("thank you for using euron calculator")
            
```
