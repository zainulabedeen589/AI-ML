```python
marks = int(input("enter your marks: "))

if marks >= 90:
    print("A+ grade")
elif marks >= 80:
    print("A grade")
elif marks >= 70:
    print("B+ grade")
elif marks >= 60:
    print("B grade")
elif marks >= 50:
    print("C grade")
else:
    print("fail")
```
