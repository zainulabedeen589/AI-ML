```python
class bankAccount:
    def __init__(self,holder_name,account_number,balance=0):
        self.holder_name = holder_name
        self.account_number = account_number
        self.balance = balance
    
    def deposit(self,amount):
        self.balance = self.balance + amount    
        return self.balance 
    
    def withdraw(self,amount):
        if amount > self.balance:
            return "Insufficient Balance"
        else:
            self.balance = self.balance - amount
            return self.balance 
    
```

```python
santanu = bankAccount("Santanu haldar", "1234567890")
```

```python
class Shoppingcart:
    customer_name = "rakkesh"
    def __init__(self,customer_name):
        self.customer_name = customer_name
        self.items = []
        
    def add_item(self,item_name,price,quantity):
        item = {
            "item_name": item_name,
            "price": price,
            "quantity": quantity
        }
        self.items.append(item)
        return self.items
    
    def total_items(self):
        return len(self.items)
    
    def total_cost(self):
        total = 0
        for item in self.items:
            total += item["price"] * item["quantity"]
        return total
```

```python
santanu = Shoppingcart("Santanu Haldar")
```

```python
santanu.add_item("apple",10,2)
```

```python
santanu.add_item("dairy milk",45,5)
```

```python
santanu.total_items()
```

```python
santanu.total_cost()
```
