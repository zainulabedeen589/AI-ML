```python
class order:
    def __init__(self,cart_total,dist_km):
        self.cart_total = cart_total
        self.dist_km = dist_km
        
    def delivery_fee(self):
        base = 20
        per_km = 5
        total_charge = per_km*self.dist_km + base
        return total_charge
    
    def fianl_amount(self):
        return self.cart_total + self.delivery_fee()
        
```

```python
class FreeDeliverypass(order):
    def delivery_fee(self):
        if self.cart_total > 1000:
            return 0
        return super().delivery_fee()
    
    
```

```python
o4 = FreeDeliverypass(500,10)
```

```python
o4.delivery_fee()
```

```python
o4.fianl_amount()
```

```python
class rainsurge(order):
    def delivery_fee(self):
        base_fee = super().delivery_fee()
        surge_amount = .25 * base_fee
        return base_fee + surge_amount
        
```

```python
o1 = order(2000,12)
```

```python
o1.delivery_fee()
```

```python
o1.fianl_amount()
```

```python
class Email:
    def send(self,message):
        print(f"this is your email {message}")
        
class sms:
    def send(self,message):
        print(f"this is form your sms {message}")
        
class notification:
    def send(self,message):
        print(f"this is form your notification {message}")
```

```python
channel = [Email(),sms(),notification()]
```

```python
def notify_all(channel, message,userinput):
    for ch in channel:
        ch.send(message)
```
