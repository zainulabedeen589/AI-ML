```python
def fectorial(n):
    if n==0 or n==1:
        return 1
    else:
        return n * fectorial(n-1)
```

```python
def fibonachicci(n):
    if n<=0:
        return "invalid input"
    elif n==1:
        return 0
    elif n==2:
        return 1
    else:
        return fibonachicci(n-1) + fibonachicci(n-2)
```

```python
fibonachicci(7)
```
